## 3 əsas çıxarış

2. Norma vektorun uzunluğunu, dot product isə iki vektorun istiqamət oxşarlığını göstərir.
3. Vahid vektor istiqaməti saxlayır, uzunluğu isə həmişə 1-dir.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([1, 2])
b = np.array([3, 4])
```

```python
plt.quiver(0, 0 ,a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0 ,b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vektorlar')
plt.legend()
plt.grid()
plt.show()
```

## Vektorların vizualizasiyası
Bu şəkildə iki vektor göstərilir. 
Mavi ox - a vektoru [1,2], qırmızı ox - b vektoru [3,4].
Hər ox orijindən (0,0) başlayır.

```python
def is_orthogonal(v1, v2):
    if np.dot(v1, v2) == 0:
        return "Ortogonaldır"
    else:
        return "Ortogoan deyil"
print(is_orthogonal(a, b))

v3= np.array([1,0])
v4= np.array([0,1])
print(is_orthogonal(v3, v4))
```

## Ortogonal yoxlama
Əgər iki vektorun dot product-i 0-dırsa, onlar ortogonaldır (90° bucaq).
v3=[1,0] və v4=[0,1] ortogonaldır, a və b isə deyil.


```python
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
v3 = np.array([7,8,9])

matrix = np.array([v1,v2,v3])
det = np.linalg.det(matrix)
print('Determinanant:', det)

if abs(det) < 1e-10:
    print('Xətti asılıdır')
else:
    print('Xətti asılı deyil')
```
