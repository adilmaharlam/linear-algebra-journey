# Bu həftə NumPy ilə vektor əməliyyatlarını öyrəndim:
- Vektor yaratmaq
- Toplama, çıxma, skalyar vurma
- Norma, dot product, bucaq
- Vahid vektor, linear combination
- 3D vektorlar

## 3 əsas çıxarış

1. NumPy vektorları Python list-dən fərqlidir — riyazi əməliyyatlar üçün optimallaşdırılıb.
2. Norma vektorun uzunluğunu, dot product isə iki vektorun istiqamət oxşarlığını göstərir.
3. Vahid vektor istiqaməti saxlayır, uzunluğu isə həmişə 1-dir.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([1, 2])
b = np.array([3, 4])
```

## Vektorların vizualizasiyası
Bu şəkildə iki vektor göstərilir. 
Mavi ox - a vektoru [1,2], qırmızı ox - b vektoru [3,4].
Hər ox orijindən (0,0) başlayır.

```python
plt.quiver(0, 0 ,a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0 ,b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vektorlar')
plt.legend()
plt.grid()
plt.show()
```

## Ortogonal yoxlama
Əgər iki vektorun dot product-i 0-dırsa, onlar ortogonaldır (90° bucaq).
v3=[1,0] və v4=[0,1] ortogonaldır, a və b isə deyil.


```python
def is_orthogonal(v1, v2):
    if np.dot(v1, v2) == 0:
        return "Ortogonaldır"
    else:
        return "Ortogoan deyil"
print(is_orthogonal(a, b))

v3= np.array([1,0])
v4= np.array([0,1])
print(is_orthogonal(v3, v4))
```

## Linear dependent
3 vektorun determinantı 0-dırsa, onlar linear dependentdir.
Bu o deməkdir ki, bir vektor digərlərinin kombinasiyası ilə ifadə oluna bilir.
v3 = v1 + v2 olduğuna görə bu vektorlar linear dependentdir.


```python
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
v3 = np.array([7,8,9])

matrix = np.array([v1,v2,v3])
det = np.linalg.det(matrix)
print('Determinanant:', det)

if abs(det) < 1e-10:
    print('Xətti asılıdır')
else:
    print('Xətti asılı deyil')
```

## Toplama vizualizasiyası
Mavi ox - a vektoru, qırmızı ox - b vektoru, yaşıl ox - cəmi (a+b).
Üç vektor orijindən başlayır, yaşıl ox onların cəmini göstərir.

```python
result_add = np.add(a, b)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.quiver(0, 0, result_add[0],result_add[1], angles='xy', scale_units='xy', scale=1, color='green', label='a+b')
plt.xlim(0,8)
plt.ylim(0,8)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vektorların Toplamı')
plt.legend()
plt.grid()
plt.show()
```

## Çıxma vizualizasiyası
Mavi ox - a vektoru, qırmızı ox - b vektoru, bənövşəyi ox - fərq (a-b).
a-b mənfi çıxdı çünki b vektoru a-dan böyükdür.



```python
result_sub = np.subtract(a, b)
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.quiver(0 ,0 , result_sub[0], result_sub[1], angles='xy', scale_units='xy', scale=1, color='purple', label='a-b')
plt.xlim(-2, 5)
plt.ylim(-2, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vektorların Fərqi')
plt.legend()
plt.grid()
plt.show()
```

## Skalyar vurma vizualizasiyası
Mavi ox - a vektoru, narıncı ox - 2*a vektoru.
Skalyar vurma vektorun istiqamətini dəyişmir, yalnız uzunluğunu artırır.


```python
result_scalar = 2 * a
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, result_scalar[0], result_scalar[1], angles='xy', scale_units='xy', scale=1, color='orange', label='2*a')
plt.xlim(0, 5)
plt.ylim(0, 10)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Skalyar vurma')
plt.legend()
plt.grid()
plt.show()
```

## Norma vizualizasiyası
Norma vektorun uzunluğudur. a=[1,2] vektorunun norması 2.24-dür.
Şəkildə vektorun uzunluğu label-də göstərilir.


```python
norm_a = np.linalg.norm(a)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'a, norma={norm_a:.2f}')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Norma')
plt.legend()
plt.grid()
plt.show()
```

## Dot product vizualizasiyası
Dot product iki vektorun istiqamət oxşarlığını göstərir.
a və b üçün dot product 11-dir — bu o deməkdir ki, vektorlar eyni istiqamətdədir.


```python
dot = np.dot(a, b)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title(f'Dot product = {dot}')
plt.legend()
plt.grid()
plt.show()
```

## Bucaq vizualizasiyası
Mavi ox - a vektoru, qırmızı ox - b vektoru.
İki vektor arasındakı bucaq 10.30 dərəcədir — bu cos θ formulu ilə hesablanıb.


```python
cos_theta = dot / (np.linalg.norm(a) * np.linalg.norm(b))
angle = np.degrees(np.arccos(cos_theta))

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title(f'Bucaq = {angle:.2f} dərəcə')
plt.legend()
plt.grid()
plt.show()
```

## Vahid vektor vizualizasiyası
Mavi ox - a vektoru, yaşıl ox - vahid vektoru.
Vahid vektor eyni istiqamətdədir, amma uzunluğu həmişə 1-dir.


```python
unit_a = a / np.linalg.norm(a)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
plt.quiver(0, 0, unit_a[0], unit_a[1], angles='xy', scale_units='xy', scale=1, color='green', label='vahid a')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vahid vektor')
plt.legend()
plt.grid()
plt.show()
```
